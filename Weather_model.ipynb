```python
import requests
import numpy as np
```

```python
def locator(city_name):
    base_url = "https://geocoding-api.open-meteo.com/v1/search"
    query_parameters = {
        "name" : city_name,
        "count" : 1

    }
    
    try :
        response = requests.get(base_url,params = query_parameters)
        locationdata= response.json()

        if response.status_code == 200  and "results" in locationdata :
            latitude = locationdata["results"][0]["latitude"]
            longitude = locationdata["results"][0]["longitude"]
            return latitude,longitude
        else:
            print(f"Could not find the {city_name} , please check your spelling")
            return None, None 
    except requests.exceptions.RequestException :
        print("Error!! Could not connect to the network")

def weather_data(latitude, longitude):
    base_url1 = "https://api.open-meteo.com/v1/forecast"
    query_parameters_1 = {
        "latitude": latitude,
        "longitude": longitude,
        "daily": ["temperature_2m_max", "rain_sum", "wind_speed_10m_max"],
        "past_days": 5,
        "forecast_days": 1,
	    "timezone": "auto",

        }  
    try: 
        response1= requests.get(base_url1,params = query_parameters_1) 
        weatherdata= response1.json()

        if response1.status_code == 200:
            max_temperature = weatherdata["daily"]["temperature_2m_max"][:5]
            rain_sum = weatherdata["daily"]["rain_sum"][:5]
            max_windspeed = weatherdata["daily"]["wind_speed_10m_max"][:5]
            return [max_temperature,rain_sum , max_windspeed]
        else:
            print("Enter the data carefully")
    except requests.exceptions.RequestException:
        print("Error!! Could not connect to the newtwork")

if __name__ == "__main__":
    cities = []
    for i in range(3):
        city_name = input(f"Enter the name of your city {i+1} ")
        cities.append(city_name)
    weather_matrix= []
    for city in cities:
        latitude, longitude = locator(city)
        if latitude and longitude:
            data = weather_data(latitude, longitude) 
            weather_matrix.append(data)         
    print(weather_matrix)
        

```

```python
print(cities)
weather_1 = np.array(weather_matrix)
print(weather_1)
```

```python
weather_actual = np.transpose(weather_1)
print(weather_actual)
```

```python
city1_temps = weather_actual[ : , 0 , 0]
print(f"Original temperatures for city 1 is {city1_temps}")
```

```python
sorted_temp1=  np.sort(city1_temps)
print(f"The max temp in Mumbai for the last five days was {sorted_temp1}")
print(f"The max temperature in the last 5 days was {sorted_temp1.max()}°C and minimum temperature was {sorted_temp1.min()}°C")
```
